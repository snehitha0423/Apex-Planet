```python
import pandas as pd

df = pd.read_excel("Cleaned_ApexPlanet_DataAnalytics_Dataset.xlsx")

df.head()
```

```python
#Check the gender column
```

```python
print(df['gender'].value_counts())
```

```python
#Calculate average sales for Male and Female
```

```python
male_sales = df[df['gender'] == 'Male']['total_sales']
female_sales = df[df['gender'] == 'Female']['total_sales']

print("Male Average Sales:", male_sales.mean())
print("Female Average Sales:", female_sales.mean())
```

```python
#Perform the statistical test
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
#Calculate 95% Confidence Interval
```

```python
import numpy as np
from scipy.stats import t

# Mean difference
mean_diff = male_sales.mean() - female_sales.mean()

# Variance
s1 = male_sales.var(ddof=1)
s2 = female_sales.var(ddof=1)

# Sample sizes
n1 = len(male_sales)
n2 = len(female_sales)

# Standard error
se = np.sqrt((s1/n1) + (s2/n2))

# Welch degrees of freedom
df_welch = ((s1/n1 + s2/n2)**2) / (
    ((s1/n1)**2)/(n1-1) +
    ((s2/n2)**2)/(n2-1)
)

# Critical value for 95% confidence
critical_value = t.ppf(0.975, df_welch)

# Confidence interval
lower = mean_diff - critical_value * se
upper = mean_diff + critical_value * se

print("Mean Difference:", mean_diff)
print("95% Confidence Interval:", (lower, upper))
```

```python
results = pd.DataFrame({
    'Metric': [
        'Male Average Sales',
        'Female Average Sales',
        'Mean Difference',
        'T-statistic',
        'P-value',
        'Significance Level',
        '95% CI Lower',
        '95% CI Upper'
    ],
    'Value': [
        male_sales.mean(),
        female_sales.mean(),
        mean_diff,
        t_stat,
        p_value,
        0.05,
        lower,
        upper
    ]
})

display(results)
```
